```python
import torch.optim as optim
import torch.nn as nn
import torch
from torch.utils.data import DataLoader
```

```python
from Flowers.dataset import load_and_transform, download_datasets
from Flowers.trainer import FlowersCNN, FlowersTrainer
```

```python
torch.manual_seed(42)
```

```python
download_datasets()
```

```python
train, val, test = load_and_transform('flower_data')
```

```python
len(train)
```

```python
len(val)
```

```python
len(test)
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
loss_func = nn.CrossEntropyLoss()
model = FlowersCNN()
optimizer = optim.Adam(model.parameters(), lr=0.0005, weight_decay=0.0005)
```

```python
train_loader = DataLoader(train, batch_size=64, shuffle=True)
val_loader = DataLoader(val, batch_size=64, shuffle=True)
test_loader = DataLoader(test, batch_size=64, shuffle=False)
```

```python
data, target = next(iter(train_loader))
print(target.max(), target.min())
print(target)
print(data.shape)
```

```python
FlowersModel = FlowersTrainer(
    model=model,
    train_loader=train_loader,
    val_loader=val_loader,
    test_loader=test_loader,
    loss=loss_func,
    optim=optimizer,
    device=device,
    epochs=35
)
FlowersModel.train_and_evaluate()
```
